#Task 4: Data Merging and Advanced Analysis


```python
# 1. Aggregate registered users by state
users_state = (
    state_txn.groupby("State", as_index=False)["Registered Users"]
    .sum()
)

# 2. Keep only one population record per district
#    This prevents duplicate district populations from being counted multiple times
demo_unique = district_demo[
    ["State", "District", "Population"]
].drop_duplicates(subset=["State", "District"])

# 3. Aggregate population by state
population_state = (
    demo_unique.groupby("State", as_index=False)["Population"]
    .sum()
)

# 4. Merge users and population
state_ratio = pd.merge(
    users_state,
    population_state,
    on="State",
    how="inner"
)

# 5. Calculate ratio
state_ratio["Users to Population Ratio"] = (
    state_ratio["Registered Users"] /
    state_ratio["Population"]
)

# 6. Calculate percentage
state_ratio["Users to Population (%)"] = (
    state_ratio["Users to Population Ratio"] * 100
)

# 7. Display final table
state_ratio[
    [
        "State",
        "Registered Users",
        "Population",
        "Users to Population Ratio",
        "Users to Population (%)"
    ]
].style.format({
    "Registered Users": "{:,.0f}",
    "Population": "{:,.0f}",
    "Users to Population Ratio": "{:.4f}",
    "Users to Population (%)": "{:.2f}%"
})
```

#4.2) correlate population density with transaction volumen
# (1) merge the district_txn and user and district demographics

```python
# Check the column names first
print(district_txn.columns.tolist())
print(district_demo.columns.tolist())

txn_data = (
    district_txn
    .groupby("District", as_index=False)["Transactions"]
    .sum()
)
print(txn_data)
```

```python
# 1. Aggregate total transactions by district
txn_data = (
    district_txn
    .groupby("District", as_index=False)["Transactions"]
    .sum()
)

# 2. Get population density for each district
demo_data = district_demo[
    ["District", "Density"]
].drop_duplicates(subset=["District"])

# 3. Rename Density for clarity
demo_data = demo_data.rename(
    columns={"Density": "Population Density"}
)

# 4. Merge transaction data with demographics
district_merged = pd.merge(
    txn_data,
    demo_data,
    on="District",
    how="inner"
)

# 5. Display merged data
print(district_merged.head())
```

#correlation between Population Density and Transaction Volume

```python
correlation = district_merged["Population Density"].corr(
    district_merged["Transactions"]
)

print("Correlation between Population Density and Transaction Volume:", correlation)
```

# create a scatter plot to show the population density correlation with Transaction volume

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))

plt.scatter(
    district_merged["Population Density"],
    district_merged["Transactions"],
    color="steelblue",
    alpha=0.6,
    edgecolors="black"
)

plt.title(
    f"Population Density vs Transaction Volume\n"
    f"Correlation = {correlation:.3f}",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Population Density")
plt.ylabel("Transaction Volume")

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#4.3) Average transaction amount per user


```python
# Total transaction amount by state
amount_state = (
    district_txn
    .groupby("State", as_index=False)["Amount (INR)"]
    .sum()
)

# Registered users by state
users_state = (
    state_txn
    .groupby("State", as_index=False)["Registered Users"]
    .max()
)

# Merge both datasets
avg_txn_user = pd.merge(
    amount_state,
    users_state,
    on="State",
    how="inner"
)

# Calculate average transaction amount per user
avg_txn_user["Average Transaction Amount per User"] = (
    avg_txn_user["Amount (INR)"] /
    avg_txn_user["Registered Users"]
)

# Display result
avg_txn_user[
    [
        "State",
        "Amount (INR)",
        "Registered Users",
        "Average Transaction Amount per User"
    ]
].style.format({
    "Amount (INR)": "₹{:,.2f}",
    "Registered Users": "{:,.0f}",
    "Average Transaction Amount per User": "₹{:,.2f}"
})
```

5.1) Line plot — Total Transactions and Amount over Time for a Selected State


```python
import matplotlib.pyplot as plt

# Select state
selected_state = "Bihar"

# Filter data for selected state
data = state_txn[state_txn["State"] == selected_state].copy()

# Create Year-Quarter column
data["Time"] = data["Year"].astype(str) + "-Q" + data["Quarter"].astype(str)

# Sort data
data = data.sort_values(["Year", "Quarter"])

# Create figure
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot transactions
ax1.plot(
    data["Time"],
    data["Transactions"],
    marker="o",
    color="blue",
    label="Transactions"
)

ax1.set_xlabel("Year and Quarter")
ax1.set_ylabel("Total Transactions", color="blue")

# Create second y-axis
ax2 = ax1.twinx()

# Plot amount
ax2.plot(
    data["Time"],
    data["Amount (INR)"],
    marker="o",
    color="green",
    label="Transaction Amount"
)

ax2.set_ylabel("Transaction Amount (INR)", color="green")

# Title
plt.title(f"Transactions and Transaction Amount Over Time - {selected_state}")

# Rotate labels
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
